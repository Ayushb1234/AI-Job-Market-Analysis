# First Dataset Insights Layoffs Companies

```python
import pandas as pd
```

```python
df = pd.read_csv("C:\\Users\\ka843\\Coding\\AI-Job-Market-Analysis\\data\\raw\\layoffs dataset\\Cleaned_tech_layoffs.csv")    
df.head()
```

```python
df.isna().sum() / len(df) * 100
```

```python
df['Country'].value_counts()
```

```python
df.info()
```

```python
df['Date_layoffs'] = pd.to_datetime(df['Date_layoffs'], errors='coerce')
```

```python
df.info()
```

```python
df['USState'] = df['USState'].fillna(df['USState'].mode()[0])
```

```python
# finding missing value in column ussate
df.isna().sum()
```

```python
df.head(5)
```

```python
df['Stage'].value_counts()
```

```python
df['year'] = df['Date_layoffs'].dt.year
```

```python
import matplotlib.pyplot as plt


yearly = df.groupby('year')['Laid_Off'].sum()
yearly.plot(kind='bar')
plt.xlabel('Year')
plt.ylabel('Number of Layoffs')
plt.title('Yearly Layoffs')
plt.show()
```

```python
# most laid off by which company
company_layoffs = df.groupby('Company')['Laid_Off'].sum().sort_values(ascending=False)
print(company_layoffs.head(10))
```

```python
top_10_companies = company_layoffs.head(10)
top_10_companies.plot(kind='bar', figsize=(10, 6), color='skyblue', edgecolor='black', width=0.8, alpha=0.7, linewidth=1.5)
plt.xlabel("Company")
plt.ylabel("No. of Companies")
plt.title("Top 10 companies with most layoffs")
plt.xticks(rotation=45)
plt.show()

# top 10 companies only


```

```python
# which country seen be the highest layoffs

df['Country'].value_counts().head(10).plot(kind='bar', figsize=(10, 6), color='salmon', edgecolor='black', width=0.8, alpha=0.7, linewidth=1.5)
plt.xlabel("Country")
plt.ylabel("No.of layoffs in thousands")
plt.title("Top 10 Countries with most layoffs")
plt.xticks(rotation=45)
plt.show()
```

```python
# which sectors are getting highest layoffs % wise use pie chart
df['Industry'].value_counts().head(10).plot(kind='pie', figsize=(8, 8), autopct='%1.1f%%', startangle=140, colors=plt.cm.Pastel1.colors)
plt.title("Top 10 Industries with most layoffs in world")
plt.axis('equal')
plt.show()
```

```python
# which sectors are getting highest layoffs % wise use pie chart in India

df_india = df[df["Country"] == "India"]
df_india = df_india['Industry'].value_counts().head(10).plot(kind='pie', figsize=(8, 8), autopct='%1.1f%%', startangle=140, colors=plt.cm.Pastel1.colors)
plt.title("Top 10 Industries with most layoffs in India")
plt.axis('equal')
plt.show()
```

```python
# per year layoffs in India
df[df["Country"] == "India"].groupby("year")["Laid_Off"].sum().plot(kind="bar", figsize=(10, 6), color="lightgreen", edgecolor="black", width=0.8, alpha=0.7, linewidth=1.5)
plt.xlabel("Year")
plt.ylabel("No. of layoffs")
plt.title("Yearly Layoffs in India")
plt.xticks(rotation=45)
plt.show()
```

```python
# company layoffs in india most top 10
df[df["Country"] == "India"].groupby("Company")["Laid_Off"].sum().sort_values(ascending=False).head(10).plot(kind="bar", figsize=(10, 6), color="lightcoral", edgecolor="black", width=0.8, alpha=0.7, linewidth=1.5)
plt.xlabel("Company")
plt.ylabel("No. of layoffs")
plt.title("Top 10 Companies with most layoffs in India (2020-2025)")
plt.xticks(rotation=45)
plt.show()
```
